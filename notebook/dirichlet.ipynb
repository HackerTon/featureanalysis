```python
def cross_entropy_dirichlet(prediction: torch.Tensor, target: torch.Tensor):
    alpha = prediction + 1
    diriclet_strength = alpha.sum(dim=-1)
    return (target * torch.digamma(diriclet_strength) - torch.digamma(alpha)).sum(
        dim=-1
    )


def gamma(tensor):
    return torch.pow(torch.e, torch.lgamma(tensor))


def KL_divergence_dirichlet(prediction: torch.tensor, target: torch.Tensor):
    n_class = torch.tensor(prediction.size(-1))
    approx_alpha = target + (1 - target) * prediction
    first_term = torch.log(
        gamma(approx_alpha.sum(dim=-1))
        / (gamma(n_class) * gamma(approx_alpha).prod(dim=-1))
    )

    second_term = (
        (approx_alpha - 1)
        * (torch.digamma(approx_alpha) - torch.digamma(approx_alpha.sum(dim=-1)))
    ).sum(dim=-1)

    return first_term + second_term


def overall_loss(
    prediction: torch.Tensor,
    target: torch.Tensor,
    lambda_t: torch.Tensor,
):
    return cross_entropy_dirichlet(
        prediction, target
    ) + lambda_t * KL_divergence_dirichlet(
        prediction,
        target,
    )


overall_loss(torch.tensor([[20, 1]]), torch.tensor([[1, 0]]), torch.tensor(0.0))
```
